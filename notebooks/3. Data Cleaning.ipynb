

## 3. Data Cleaning

Data cleaning was performed to improve data quality and ensure that the dataset is suitable for further analysis and machine learning.

The following operations were performed:

1. **Duplicate Removal:** Removed 34,335 duplicate records using `drop_duplicates()`.

2. **Missing Values:** Filled missing product descriptions with `"Unknown"`. Missing Customer IDs were retained because they are important identifiers for customer-level analysis and cannot be reliably imputed.

3. **Data Type Conversion:**
   - Converted `InvoiceDate` into datetime format.
   - Converted `Customer ID` into Pandas nullable integer format (`Int64`).

4. **Feature Creation:** Created a new column `TotalAmount` by multiplying `Quantity` and `Price`.

5. **Negative Quantity Handling:** Removed 3,393 negative-quantity records that were not associated with cancellation invoices. Retained cancellation records for transaction-history analysis.

6. **Zero Price Removal:** Removed transactions with zero unit price because the project focuses on predicting future paid purchases.

7. **Negative Price Handling:** Removed 5 records with negative prices, identified as `Adjust bad debt` accounting adjustments.

The cleaned dataset was saved separately as `online_retail_working.csv`, preserving the original combined dataset.



## 4. Exploratory Data Analysis (EDA)

Exploratory Data Analysis is performed to understand the underlying patterns, distributions, relationships and characteristics of the cleaned dataset before applying machine learning algorithms.

### Objectives of EDA

- Understand individual variable distributions.
- Identify outliers and unusual patterns.
- Examine relationships between different variables.
- Discover customer purchasing behaviour.
- Identify important patterns that can help in feature engineering and model development.

### EDA Techniques

1. **Univariate Analysis:** Study individual variables using descriptive statistics, frequency distributions and visualizations.

2. **Outlier Analysis:** Identify unusual observations using methods such as IQR and box plots. Investigate whether they are genuine observations or errors.

3. **Bivariate Analysis:** Examine relationships between two variables using suitable plots and statistical methods.

4. **Multivariate Analysis:** Study relationships among multiple variables to identify complex patterns.

5. **Correlation Analysis:** Examine the strength and direction of relationships between numerical variables.

6. **Statistical Analysis:** Apply suitable hypothesis tests and confidence intervals to validate findings.

The findings from EDA will guide feature engineering and the selection of relevant variables for the machine learning models.
