5. Negative Quantity:

6. Negative Price:
   - Investigate 5 negative price records.

   - Separate accounting adjustments from normal product transactions.

7. Zero Price:
   - Investigate 6,202 zero-price records.
   - Identify free products, promotional items and unusual transactions.

8. Outliers:
   - Investigate extreme Quantity and Price values.
   - Avoid blindly deleting legitimate bulk purchases or unusual transactions.

9. Cancelled Transactions:

   - Analyse their relationship with negative quantities and returns.
